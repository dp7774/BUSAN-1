# 2. 각 항목의 기여 정도:환적 편
전체 물동량을 수출, 수입, 환적 등의 요소로 나누어 각각이 전체 성장에 얼마나 기여 했는지를 살펴볼 수 있음.
예시)
전체 컨테이너 물동량 2000만 -> 2400만 증가
수출입 물동량 1000만 -> 1050만
환적 1000만 -> 1350만
== 전체 증가량의 대부분을 환적이 차지

1. 환적 물동량의 전반적인 증가량
2. 전체 물동량에서 환적이 차지하는 비중

```python
import pandas as pd
import numpy as np
```

```python
growth = pd.read_csv(
    r'C:\practice1\pr1\trans_table.csv',
    encoding='cp949',
    index_col='연도'
)

growth.head()
```

```python
pd.options.display.float_format = '{:,.2f}'.format

growth.head()
```

```python
growth['전체증감'] = growth['전체물동량'].diff()

growth['환적증감'] = growth['환적'].diff()

growth.head(10)
```

```python
growth['환적기여도'] = (
    growth['환적증감']
    / growth['전체증감']
    * 100
).round(2)

growth.head(25)
```

| 전체증감 | 환적증감 | 의미                   |
| ---: | ---: | -------------------- |
|    + |    + | 환적이 전체 **성장에 기여**    |
|    + |    - | 환적이 전체 **성장을 억제**    |
|    - |    - | 환적이 전체 **감소에 기여**    |
|    - |    + | 환적 증가가 전체 **감소를 완화** |


```python
growth_year = growth[
    (growth['전체증감'] > 0) &
    (growth.index <= 2024)
].copy()

growth_year
# # 2005년 110%는 오류가 아니라, 수출입 물동량이 감소했음에도 환적 증가가 이를 상쇄했다는 해석
```

```python
high_contribution = growth_year[
    growth_year['환적기여도'] >= 50
].copy()

high_contribution
```

| 지표    | 질문                                |
| ----- | --------------------------------- |
| 환적비율  | 전체 물동량 중 환적이 얼마나 차지하는가?           |
| 환적기여도 | 전체 물동량이 증가한 양 중 환적 증가가 얼마나 차지했는가? |
