```python
df = df.sort_values(['연도', '월'])

df.head(20)
```

# (1) 장기성장

```python
yearly = df.groupby('연도')['물동량'].sum() 
# 같은 연도끼리의 데이터를 묶는다.
# ['물동량'].sum() -> 물동량을 모두 더한다.

yearly 
```

```python
yearly_growth = yearly.pct_change() * 100
# pct_change(): 바로 앞에 값과 비교해서 얼마나 변했는지 계산

yearly_growth
```

+ 음수 값 빼서 정리
+ 최고 성장 연고 / 최저 성장 연도

# (3) 환적구조 파악

```python
df['수출입구분설명'].unique()
```

```python
trade_type = df.groupby('수출입구분설명')['물동량'].sum()

trade_type
```

```python
trade_ratio = trade_type / trade_type.sum() * 100

trade_ratio
```
