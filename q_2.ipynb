```python
import pandas as pd
import numpy as np
```

```python
CSV_PATH = './D001.csv'
df = pd.read_csv(CSV_PATH, encoding='cp949')
```

```python
data = df.copy()
data
```

```python
data = data.sort_values(["연도", "월"])
data
```

```python
data["구조"] = data["수출입구분설명"].replace({
    "수입환적": "환적",
    "수출환적": "환적",
    "수입": "수출입",
    "수출": "수출입"
})
data
```

```python
year_structure = (
    data.groupby(["연도", "구조"])["물동량"]
    .sum()
    .unstack()
    .reset_index()
)

year_structure
```

```python
# 전년 대비 환적 물동량 증가량
year_structure["환적증가량"] = (
    year_structure["환적"].diff()
)

year_structure
```

```python
year_structure["환적기여도"] = (
    year_structure["환적증가량"] * 100
)

year_structure
```
